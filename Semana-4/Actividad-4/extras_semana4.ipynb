Extra 1: Suma de todos los elementos de una matriz 

```python
matriz = [["1","2","3"],
          ["4","5","6"],
          ["7","8","9"]
]

def suma_matriz(matriz):
    total = 0
    for i in matriz: #Delimita el recorrido 
        for j in i:
             total += int(j) # Convierte las variables de la matriz a int para poder sumarse 

    return (f"La suma de todos los elementos es: {total}")

         
def imprimir_matriz(matriz):
    for f in matriz: 
        for e in f: 
            print(e, end=" ")
        print("") 


imprimir_matriz(matriz)
suma_matriz(matriz)
```

Extra 2: Matriz con suma por fila 

```python
matriz = [[3, 1, 4], [1, 5, 9], [2, 6, 5]]
fila = 0
def suma_fila(fila):
    total = 0
    for elemento in fila: # recorre toda la fila seleccionada 
        total += elemento # Suma todos los valores de la fila
    return total
def mostrar_tabla(matriz):   #          Imprime
    for f in matriz:         #            toda
        for e in f:          #             la
            print(e, end=" ")#           matriz
        print("")            # Sin necesidad de un return

mostrar_tabla(matriz)
for i in range(len(matriz)):
    print(f"Suma de la fila {i}: {suma_fila(matriz[i])}")
```

Extra 3: Producto de una columna sin usar el operador * 

```python
matriz = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9],
          [2, 4, 6]
]
columna = int(input("Ingresar num de columna que quiera sumar(0,1 o 2):"))
 
def multiplicar_columna(matriz, columna):
    total = 1
    for fila in matriz: 
        valor = fila[columna] # Ayuda a identificar cuales valores en la matriz se sumarán 
        resultado = 0
        for s in range(valor): #s se toma como la suma constante dentro del rango
            resultado += total
        total = resultado 
    print(f"El producto de la columna {columna} es: {total}")
    return 

multiplicar_columna(matriz, columna)
```

# Preguntar cuantas calificaciones desea promediar y promedialas

```python
def promedio(sumatoria, n):
    return sumatoria / n


n = int(input("Cuántas calificaciones quieres promediar?"))
sumatoria = 0

print(f"Se van a pedir {n} calificaciones")
for c in range(1,n + 1):
    calif = float(input(f"Dame la calificación {c}"))
    sumatoria = sumatoria + calif 

print(f"El promedio de {n} calificaciones es {promedio(sumatoria, n)}")  
```

```python
filas = 3 
columnas = 3

matriz = []
for f in range(filas):
    fila_actual = []
    for c in range(columnas):
        # matriz[f][c] = int(input(f"Dame valor para [{f}][{c}]: "))
        valor = int(input(f"Dame valor para [{f}][{c}]: "))
        fila_actual.append(valor)
    matriz.append(fila_actual)

print("Impresión de la matriz")
for f in range(filas):
    for c in range(columnas):
        print(matriz[f][c], end=" ")
    print()     


```

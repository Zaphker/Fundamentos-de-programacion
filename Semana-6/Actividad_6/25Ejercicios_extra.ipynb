### Ejercicio 1: Modelo EPS del área de un triángulo

```python
print("------ Area de un Triangulo ------")
base = float(input("Ingresar la base del triangulo: "))
altura = float(input("INgresar la altura del triangulo: "))
area = (base * altura) / 2 

print(f"El area del triángulo es: {area}")
```

### Ejercicio 9: Mayor de tres números 

```python
num1 = int(input("Ingresa el número 1:"))
num2 = int(input("Ingresa el número 2:"))
num3 = int(input("Ingresa el número 3:"))

if num1 == num2 and num1 > num3:
    print(f"El mayor de los tres números es: {num1}")
elif num2 == num1 and num2 > num3:
    print(f"El mayor de los tres numeros es: {num2}")
elif num3 == num2 and num3 > num1:
    print(f"El mayor de los tres numeros es: {num3}")
elif num2 == num1 and num2 < num3:
    print(f"El mayor de los tres números es: {num3}")
else: 
    print("Todos los números son iguales.")
```

### Ejercicio 12: Suma de pares del 1 al 20 con `for` y `continue`

```python
suma = 0
for numero in range(1, 21):
    if numero % 2 != 0:
        continue
    suma += numero

print(f"La suma de los números pares del 1 al 20 es: {suma}")
```

### Ejercicio 13: Suma de una lista con función que regresa valor 

```python
def sumar_lista(lista):
    suma = 0
    for e in lista:
        suma += e
    return suma

lista = [10, 20, 30, 40]
print(f"La suma de los elementos es: {sumar_lista(lista)}")
```

### Ejercicio 14: Máximo de una lista con función que regresa valor

```python
def maximo_lista(lista):
    maximo = lista[0]
    for e in lista:
        if e > maximo:
            maximo = e
    return maximo

lista = [7, 3, 9, 21]
print(f"El elemento máximo es: {maximo_lista(lista)}")
```

### Ejercicio 15: Matriz y suma de la diagonal principal 

```python
matriz = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
suma = 0

for i in range(len(matriz)):
    for j in range(len(matriz[i])):
        if i == j:
            suma += matriz[i][j]

print(f"La suma de la diagonal principal es: {suma}")
```

### Ejercicio 16: Promedio con función que no regresa valor

```python
def imprimir_promedio(calificaciones):
    promedio = sum(calificaciones) / len(calificaciones)
    print(f"El promedio es: {promedio:.2f}")

calificaciones = [8, 9, 10, 7]
imprimir_promedio(calificaciones)
```

### Ejercicio 18: Factorial con función recursiva

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

numero = 5
print(f"El factorial de {numero} es: {factorial(numero)}")
```

### Ejercicio 19: Tupla: acceso y suma con función 

```python
def sumar_tupla(tupla):
    suma = 0
    for e in tupla:
        suma += e
    return suma

numeros = (4, 8, 15, 16, 23, 42)
print(f"El tercer elemento es: {numeros[2]}")
print(f"La suma de la tupla es: {sumar_tupla(numeros)}")
```

### Ejercicio 21: Diccionario de calificaciones y promedios

```python
calif = {
    "Ana": [8, 9, 10],
    "Beto": [7, 7, 8]
}

for nombre in calif:
    promedio = sum(calif[nombre]) / len(calif[nombre])
    print(f"{nombre}: {promedio:.2f}")
```

# ACTIVIDAD EVALUABLE 5 — CERTIFICACIÓN DE CISCO NETWORKING ACADEMY

**Curso:** Solución de problemas con programación computacional
**Semana:** 6 · Repaso integral (Temas 1 al 16) **Ponderación:** 6% de
la calificación total del curso **Fechas límite:** 18 de septiembre de
2026 (Periodo I) · 20 de noviembre de 2026 (Periodo II) **Entrega:**
Viernes por la noche

------------------------------------------------------------------------

## 1. Descripción del reto

Registrarse formalmente en el portal externo de **Cisco Academy (*Skills
for All with Cisco*)**, completar el avance formativo del curso oficial
**Fundamentos de Python 1** (*Python Essentials 1*) y **acreditar** el
examen de certificación **CISCO Fundamentos de Python 1 - Examen de
Sección**, obteniendo la **insignia digital** correspondiente. La
evidencia se entrega en un documento académico de reporte con las
capturas de pantalla requeridas.

Esta actividad consolida los **Temas 1 al 16** del curso: los contenidos
del curso externo cubren, de forma alineada, algoritmos y modelo
Entrada-Proceso-Salida, variables, tipos de datos y operadores,
entrada/salida, decisiones, ciclos, listas, matrices, funciones, tuplas,
diccionarios, excepciones y strings.

------------------------------------------------------------------------

## 2. Requerimientos académicos obligatorios (agenda oficial)

1.  **Registro en el portal:** Registrarse formalmente en el portal de
    Cisco Academy (*Skills for All with Cisco*), creando y verificando
    una cuenta personal.
2.  **Avance formativo completo:** Completar todas las **lecturas**,
    **ejercicios interactivos** y **laboratorios de práctica**
    correspondientes al curso oficial **Fundamentos de Python 1** (la
    plataforma registra el avance en la bitácora de actividades).
3.  **Acreditación del examen:** Acreditar formalmente el examen **CISCO
    Fundamentos de Python 1 - Examen de Sección**, alcanzando la
    **calificación oficial aprobatoria** que la plataforma establece.
4.  **Insignia digital:** Obtener la **insignia digital** que otorga la
    plataforma al acreditar el examen.

------------------------------------------------------------------------

## 3. Estructura del entregable (documento de reporte)

Un **documento académico de reporte** (formato Word `.docx` o PDF) que
contenga las siguientes secciones:

| \#  | Sección                   | Detalle                                                                                                                                                                                    |
|:--------------------------:|:---------------------|:---------------------|
|  1  | **Portada**               | Nombre del estudiante, matrícula, nombre del curso, institución y fecha de entrega.                                                                                                        |
|  2  | **Registro en el portal** | Descripción breve del registro y **captura de pantalla** del perfil o panel (*Dashboard*) de la cuenta en Cisco Skills for All.                                                            |
|  3  | **Bitácora de avance**    | **Captura de pantalla de la bitácora o reporte de actividades** de la plataforma externa que demuestre el avance de las lecturas, ejercicios y laboratorios completados previos al examen. |
|  4  | **Calificación oficial**  | **Captura de pantalla nítida** que muestre la **calificación oficial aprobatoria** del examen de sección.                                                                                  |
|  5  | **Insignia digital**      | **Captura de pantalla nítida** de la **insignia digital obtenida** (junto al nombre del estudiante).                                                                                       |
|  6  | **Conclusión**            | Reflexión breve (3 a 5 líneas) sobre cómo la certificación refuerza los Temas 1 al 16 del curso.                                                                                           |

**Notas de las capturas:** - Todas las capturas deben ser **nítidas y
legibles** (sin recortes que oculten datos clave, sin desenfoque). -
Debe verificarse visualmente que la captura de la calificación muestra
un resultado **aprobatorio** y que la insignia corresponde al curso
**Fundamentos de Python 1** (*Python Essentials 1*).

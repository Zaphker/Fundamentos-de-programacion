## 1. Descripción de la actividad

La **Presentación Final o Examen** es la actividad de cierre del curso
que consolida la totalidad de los **20 temas** estudiados. El estudiante
debe demostrar que:

1.  Comprende los fundamentos conceptuales (algoritmos, modelo EPS,
    pseudocódigo y PSeInt).
2.  Domina la programación en Python desde las entradas y salidas
    simples hasta el manejo avanzado de estructuras de datos, funciones,
    strings y archivos.
3.  Aplica el control de errores con excepciones y la depuración
    técnica.
4.  Comunica y defiende el **proyecto final integrador** entregado en la
    Semana 7 frente al grupo.

La actividad se compone de **dos modalidades complementarias** definidas
por el docente: la **presentación del proyecto final integrador frente
al grupo** y un **examen teórico-práctico de los temas del curso**. No
se introducen contenidos nuevos: todo lo evaluado pertenece al repaso y
consolidación de los Temas 1 al 20.

------------------------------------------------------------------------

## 2. Estructura de la actividad (modalidades)

### 2.1 Presentación del proyecto final integrador (frente al grupo)

El estudiante expone el proyecto desarrollado en las Semanas 3 y 7, con
una duración sugerida de **10 a 15 minutos**:

1.  **Portada y contexto:** nombre, matrícula, organización analizada y
    área de impacto.
2.  **Problema y reglas de negocio:** qué problema resuelve la solución
    y bajo qué restricciones opera.
3.  **Diseño algorítmico:** modelo Entrada-Proceso-Salida, diagrama de
    flujo y pseudocódigo de PSeInt.
4.  **Solución técnica:** menú principal como matriz, captura de fecha
    en tupla, persistencia en archivos de texto, control de excepciones
    y depuración con PDB.
5.  **Demostración en vivo:** ejecución del programa mostrando cada
    opción del menú, la lectura/escritura de archivos y al menos un caso
    de error controlado.
6.  **Conclusiones:** aprendizajes, dificultades superadas y mejoras
    futuras.

### 2.2 Examen teórico-práctico de los Temas 1 al 20

Examen que combina **preguntas teóricas** (algoritmos, EPS, PSeInt,
tipos de datos, excepciones) con **problemas prácticos de programación**
que integran decisiones, ciclos, listas, funciones, diccionarios,
strings, archivos y excepciones. La resolución práctica se entrega en
archivos `.py` o en un Jupyter Notebook.

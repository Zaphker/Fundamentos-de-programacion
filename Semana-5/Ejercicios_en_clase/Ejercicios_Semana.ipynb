```python

def suma_tupla(t):
    total = 0
    for e in t:
        total += e 
    return total

calorias = (800, 950, 700, 1300, 800, 900, 650)  
resultado = suma_tupla(calorias)
print(f"Le metiste a tu cuerpo: {resultado} calorias, gordo.")  
```

```python
contactos = {
    "Mateo": "555-3509",
    "Angel": "442-4929",
    "Carla": "55-67942"
}

print(type(contactos))
print(len(contactos)) #El interprete lee los contactos de manera corrida, en línea recta

print(contactos.get("Angel"))
print(contactos.get("Chris", "Sin registro"))
print(contactos.get("Pedro", "Sin registro"))

contactos["Chris"] = "442-3566" #Si no existe en el diccionario el interprete lo agrega
contactos["Angel"] = "447-3721" #Si existe en el diccionario pero cambia el dato, también lo modifica 
print(contactos)
```

```python
print(contactos.keys())
print(contactos.values())
print(contactos.items())
```

```python
print("Impresion de llaves:")
for n in contactos.keys():
    
    print(f" -{n}")
print()
print("-" * 15)
print()
print("Impresion de valores:")
for v in contactos.values():
    print(f" -{v}")
print()
print("-" * 15)
print()
print("Impresion Completa: ")    
for n, t in contactos.items():
    print(f"-Nombre:{n}\n  Teléfono: {t}")
```

```python
def buscar_contacto(contactos, nombre):
    return contactos.get(nombre, "Contacto No Encontrado.")

nombre = input("Escribe el nombre del contacto que buscas: ").capitalize()
telefono = buscar_contacto(contactos, nombre)

print(f"El teléfono de {nombre} es {telefono}")
```

```python
candidatos = {
    "A": 0,
    "B": 0,
    "C": 0 
}

cantidad_votos = int(input("¿Cuántos votos van a ser? "))
conteo = 1

while conteo <= cantidad_votos:
    voto = input(f"Ingresa el voto {conteo} (A, B o C): ")
    if voto.upper() in candidatos:
        print(f"Voto válido para {voto.upper()}\n")
        candidatos[(voto.upper())] += 1
        conteo += 1
    else:
        print("El voto no es válido, vota nuevamente")

print("\nResultados:")
for k, v in candidatos.items():
    print(f"Opción {k}, votos: {v}")
```

### Extra 1: Sistema de calificaciones con tuplas 

```python
calificaciones = (7.5, 9.98, 8.75, 6.5, 10.0)

nueva_calificacion1 = float(input("Agrega una nueva calificación: "))
nueva_calificacion2 = float(input("Agrega una nueva calificación: "))
calificacion3 = 8.0
nuevas_calificaciones = calificaciones + (nueva_calificacion1, nueva_calificacion2, calificacion3)
lista = list(nuevas_calificaciones)
lista.sort()
total = 0
for s in lista:
    total += s

print(f"Tercera calificación: {calificacion3}")
print(f"Nuevas Calificaciones: {nuevas_calificaciones}")
print(f"\nCalificaciones ordenadas (Mayor a menor): {lista}")
print(f"\nSuma Total: {total}")
```

### Extra 2: Agenda de contactos con búsqueda 

```python
agenda = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}
nuevo_contacto = input("Añade un nuevo contacto: ")
nuevo_telefono = input("Añade el teléfono del nuevo contacto: ")
nombre = input("Qué contacto deseas buscar? ")
agenda[nuevo_contacto] = nuevo_telefono
print("Agenda de contactos: ")
print("-" * 18)
for c, t in agenda.items(): #La "c" se encarga de buscar en los contactos 
                            #la "t" se encarga de buscar en los teléfonos 
    print(f"Nombre: {c}\n -Teléfono: {t}")
print("-" * 18)
def buscar_telefono(agenda, nombre): #Función que busca el nombre y te regresa el teléfono
    return agenda.get(nombre, "Contacto no encontrado.")

telefono = buscar_telefono(agenda, nombre)
print(f"\nEl teléfono de {nombre} es: {telefono}")
```

### Extra 3: Calculadora segura con manejo de excepciones 

```python
num1 = int(input("Ingresar número 1: "))
num2 = int(input("Ingresar número 2: "))
try:
    resultado = num1 / num2 
    print(f"{num1} dividdo entre {num2} es: {resultado}")
except ValueError:
      print("ERROR: Caracteres no númericos o vacíos.")

except ZeroDivisionError: 
     print("ERROR: No es posible dividir entre cero. Ingresa un divisor distinto de 0.")

```

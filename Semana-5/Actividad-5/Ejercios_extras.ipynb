### Extra 1: Sistema de calificaciones con tuplas 

```python
calificaciones = (7.5, 9.98, 8.75, 6.5, 10.0)

nueva_calificacion1 = float(input("Agrega una nueva calificación: "))
nueva_calificacion2 = float(input("Agrega una nueva calificación: "))
calificacion3 = 8.0
nuevas_calificaciones = calificaciones + (nueva_calificacion1, nueva_calificacion2, calificacion3)
lista = list(nuevas_calificaciones)
lista.sort()
total = 0
for s in lista:
    total += s

print(f"Tercera calificación: {calificacion3}")
print(f"Nuevas Calificaciones: {nuevas_calificaciones}")
print(f"\nCalificaciones ordenadas (Mayor a menor): {lista}")
print(f"\nSuma Total: {total}")
```

### Extra 2: Agenda de contactos con búsqueda 

```python
agenda = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}
nuevo_contacto = input("Añade un nuevo contacto: ")
nuevo_telefono = input("Añade el teléfono del nuevo contacto: ")
nombre = input("Qué contacto deseas buscar? ")
agenda[nuevo_contacto] = nuevo_telefono
print("Agenda de contactos: ")
print("-" * 18)
for c, t in agenda.items(): #La "c" se encarga de buscar en los contactos 
                            #la "t" se encarga de buscar en los teléfonos 
    print(f"Nombre: {c}\n -Teléfono: {t}")
print("-" * 18)
def buscar_telefono(agenda, nombre): #Función que busca el nombre y te regresa el teléfono
    return agenda.get(nombre, "Contacto no encontrado.")

telefono = buscar_telefono(agenda, nombre)
print(f"\nEl teléfono de {nombre} es: {telefono}")
```

### Extra 6: Contador de votos con diccionario 

```python
votos = {"Rojo": 0, "Azul": 0, "Verde": 0}

for i in range(5):
    color = input("Ingresa el color de tu voto: ")
    votos[color] += 1

for color, total in votos.items():
    print(color, ":", total)

if votos["Rojo"] >= votos["Azul"] and votos["Rojo"] >= votos["Verde"]:
    print(f"El color ganador es Rojo con {votos['Rojo']} votos")
elif votos["Azul"] >= votos["Rojo"] and votos["Azul"] >= votos["Verde"]:
    print(f"El color ganador es Azul con {votos['Azul']} votos")
else:
    print(f"El color ganador es Verde con {votos['Verde']} votos")
```

### Extra 7: Validación robusta de entrada 

```python
numero1 = None
while numero1 is None:
    try:
        numero1 = int(input("Ingresa el primer número: "))
    except ValueError:
        print("Entrada inválida, intenta de nuevo")

numero2 = None
while numero2 is None:
    try:
        numero2 = int(input("Ingresa el segundo número: "))
    except ValueError:
        print("Entrada inválida, intenta de nuevo")

try:
    resultado = numero1 / numero2
    print(f"El resultado de la división es: {resultado}")
except ZeroDivisionError:
    print("No se puede dividir entre cero")
```

### Extra 9: Minimenú modular integrador 

```python
numeros = (5, 2, 8, 1, 9)
contactos = {"Ana": "555-1234", "Luis": "555-5678", "Carla": "555-9012"}

def mostrar_tupla():
    lista_ordenada = sorted(numeros)
    print("Tupla ordenada:", lista_ordenada)

def buscar_telefono():
    nombre = input("Ingresa el nombre a buscar: ")
    if nombre in contactos:
        print(f"El teléfono de {nombre} es {contactos[nombre]}")
    else:
        print("Contacto no encontrado")

def dividir_numeros():
    try:
        num1 = int(input("Ingresa el primer número: "))
        num2 = int(input("Ingresa el segundo número: "))
        print("Resultado:", num1 / num2)
    except ZeroDivisionError:
        print("No se puede dividir entre cero")
    except ValueError:
        print("Debes ingresar números válidos")

def analizar_mensaje():
    mensaje = input("Ingresa un mensaje: ")
    print("Longitud del mensaje:", len(mensaje))
    print("En mayúsculas:", mensaje.upper())
    print("Texto reemplazado:", mensaje.replace("Python", "programación"))
    print("Palabras totales:", len(mensaje.split()))

opcion = 0
while opcion != 5:
    print("\n1. Mostrar tupla ordenada")
    print("2. Buscar teléfono en contactos")
    print("3. Dividir dos números")
    print("4. Analizar mensaje de texto")
    print("5. Salir")

    opcion = int(input("Elige una opción: "))

    if opcion == 1:
        mostrar_tupla()
    elif opcion == 2:
        buscar_telefono()
    elif opcion == 3:
        dividir_numeros()
    elif opcion == 4:
        analizar_mensaje()
    elif opcion == 5:
        print("Programa finalizado")
    else:
        print("Opción no válida")
```

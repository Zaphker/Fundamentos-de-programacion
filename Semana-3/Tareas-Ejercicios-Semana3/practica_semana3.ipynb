Ejercicio 5: Número invertido

```python
numero = int(input("Ingresar número: "))
numero_original = numero
invertido = 0

while numero != 0: #Siempre y cuando el numero no sea menor a 0 el ciclo continuará ejecutandóse
    r = numero % 10 #Se divide el número entre 10 y se convierte en el residuo
    invertido = invertido * 10 + r
    numero = numero // 10
print(f"El número invertido de {numero_original} es {invertido}")
```

Ejercicio 6: División con restas sucesivas

```python
divisor = int(input("Ingresa un divisor(Exceptuando el 0)"))
dividendo = int(input("Ingresa un número mayor a el divisor)"))

if divisor <= 0:
    print("El numero debe ser mayor a 0")
else:
    cociente = 0
    residuo = dividendo

    while residuo >= divisor:
        residuo = dividendo - divisor
        cociente = cociente + 1

print(f"{dividendo} / {divisor} = {cociente} con residuo de {residuo}")
print("Termino el proceso.")
```

Ejercicio 15: Piramide Centrada

```python
altura = int(input("ingresar el tamaño de la piramide deseada: "))

for i in range(1, altura +1):
    for j in range(1, 2 * altura):
        if j <= altura - i or j > altura - i + (2 * i - 1):
            print(" ", end="")
        else: 
            print("*", end="")
    print()
```

Ejercicio 16: Adivina el número secreto

```python
numero_secreto = 42
intentos = 0
adivinado = False
while not adivinado:
    numero_usuario = int(input("Adivina el número secreto: "))
    intentos += 1
    if numero_usuario > numero_secreto:
        print("Muy alto")
    elif numero_usuario < numero_secreto:
        print("Muy bajo")
    else:
        print(f"¡Correcto! Lo lograste en {intentos} intentos")
        adivinado = True
```

Ejercicio 17: Detector de números primos 

```python
numero = int(input("Ingresar número entero: "))

contador = 0 

for i in range (2, numero -1 ):
    if numero % i == 0:
        contador += 1

if contador == 0:
    print(f"{numero}, es primo")
else:
    print(f"{numero}, no es primo")
```

Ejercicio 18: Número Perfecto

```python
numero = int(input("Ingresar número entero:"))
suma = 0 

for i in range(1, numero):
    if numero % i == 0:
        suma += 1
if suma == numero:
    print(f"{numero} es un número perfecto")
else:
    print(f"No es un número perfecto")
```

Ejercicio 19: Interés compuesto por año

```python
capital = int(input("Ingresar capital a invertir: "))
años = int(input("¿Por cuántos años quieres invertirlo? "))
tasa = int(input("Porcentaje de intereses por año: "))

for i in range(1, años, +1):
    capital = capital * (1 + (tasa / 100))
    años += 1
    print(f"Año {i}:${capital:.2f}")

print(f"Capital final: ${capital:.2f}")
```

Ejercicio 20: Simulador de batería

```python
import time

bateria = 100
hora = 0

while bateria > 20: 
    bateria -= 10
    hora += 1
    print(f"Hora: {hora}: {bateria}%")
    time.sleep(10)


print("¡Bateria baja! Conecta el cargador")
```

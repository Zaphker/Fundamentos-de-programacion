# Avance Proyecto Final (Fase I)

---

**Análisis Organizacional:**

- _Empresa:_ Campus TecMilenio Querétaro

- _Area de Impacto Operativo_: Cafeteria General y organización en pedidos. 

- _Necesidad:_ En las cajas de cobro y organización de pedidos se hace todo mediante un proceso de hojas de papel y gritar a los que están en cocina a la hora de hacer un pedido, el area de mejora es implementar un programa mediante python para hacer todo más automatizado. 

---

**Definición del Problema:**

- La organización y eficiencia a la hora de realizar una orden para insumos es rápida pero tiene problemas cuándo se acumula la gente debido a que tienen que tomar la orden y luego llevarla a cocina, cuándo se podría simplemente tomar la orden mediante un sistema programado y que le llegue a cocina directamente para seguir tomando órdenes sin interrupciones. 

---

**Listado de Requerimientos**

- _Registrar orden por estudiante:_ El sistema debe registrar la orden completa del estudiante.

- _Definir si el pedido necesita tiempo de espera:_ Ya que hay comidas que están preparadas verificar si está disponible para entregar. 

- _Tener un menu para agregar o eliminar articulos:_ Guardar los productos y modificar para verificar si están disponibles o no.

- _Imprimir los pedidos a cocina después de terminar una orden:_ Que cocina reciba el último pedido registrado para comenzar la preparación. 

- _Suma de total por pedido:_ El sistema debe de sumar todos los articulos agregados al pedido para dar el total.

- _Registro de pedido por nombre del estudiante:_ Ingresar el nombre del estudiante para identificar las ordenes.

---

**Clasificación de Datos**

-  `nombre_estudiante` = str Nombre del estudiante a tomar orden.

- `articulo` = str =

- `precio_articulo` = int = Para agregar o modificar el precio al producto.

- `total_pedido` = int = Total de la cuenta después de la suma de todos los articulos agregados. 

- `preparado` = boolean = Saber si un articulo está listo para la venta o requiere preparación.





---

**Operadores de Lenguaje**

- Operadores matematicos usados = `+`, `*` La suma servirá a la hora de saber el total de una cuenta, la multiplicación será necesaria en caso de que sean varios articulos del mismo para multiplicar por la cantidad total

- Operadores Relacionales usados = `==`, `<=`, `>=` "Igual a" será usado para identificar los articulos seleccionados, "Menor o igual a" será usado cuándo se termine la toma del menu, "Mayor o igual a " será usado para los bucles 

- Operadores lógicos usados = `and` `or` `not` Se usarán principalmente para las variaciones en las que el estudiante decida o no tomar, pedir o cancelar un articulo, y su resulado sea necesario para una modificación 
---

**Estructuras de Control**

- `if` `elif` `else` Control de bucles para la toma de ordenes, para casos en los que sean mas de 2 articulos, este preparado o no un producto o si se necesite agregar algo mas al menu

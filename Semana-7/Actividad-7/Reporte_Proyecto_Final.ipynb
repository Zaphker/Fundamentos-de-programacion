# Reporte Final Proyecto Final

---

**Análisis Organizacional:**

- _Empresa:_ Campus TecMilenio Querétaro

- _Area de Impacto Operativo_: Cafeteria General y organización en pedidos.

- _Necesidad:_ En las cajas de cobro y organización de pedidos se hacía todo mediante un proceso de hojas de papel y gritar a los que están en cocina a la hora de hacer un pedido, el area de mejora fue implementar un programa mediante python para hacer todo más automatizado.

---

**Definición del Problema:**

- La organización y eficiencia a la hora de realizar una orden para insumos era rápida pero tenía problemas cuándo se acumulaba la gente, debido a que había que tomar la orden en papel y luego llevarla a cocina. El programa desarrollado resuelve esto tomando la orden desde un sistema programado y guardándola directamente en un archivo que cocina puede consultar, permitiendo seguir tomando órdenes sin interrupciones.

---

**Listado de Requerimientos**

- _Inicio de sesión del cajero:_ El sistema pide usuario y contraseña antes de dejar operar la caja, validando contra los datos guardados en `usuarios.txt`. La contraseña se escribe mostrando asteriscos en vez del texto real.

- _Registrar orden por estudiante:_ El sistema registra la orden completa del estudiante, navegando el menú por categorías.

- _Límite de artículos por pedido:_ No se pueden agregar más de 10 artículos en una misma orden, para evitar que se saturen los pedidos hacia cocina.

- _Definir si el pedido necesita tiempo de espera:_ Ya que hay comidas que están preparadas, el sistema verifica si el pedido completo está listo para entregar de inmediato o si requiere preparación.

- _Tener un menú para agregar o eliminar artículos:_ El menú se guarda en `menu.txt` y se puede modificar en cualquier momento, agregando o eliminando artículos y su categoría.

- _Imprimir los pedidos a cocina después de terminar una orden:_ Cada pedido terminado se muestra en pantalla como ticket y se guarda de forma permanente en `pedidos.txt` para que cocina lo consulte.

- _Suma de total por pedido:_ El sistema suma todos los artículos agregados al pedido, considerando cantidad y precio, para dar el total a pagar.

- _Registro de pedido por nombre del estudiante:_ Se pide el nombre del estudiante al iniciar cada orden, para identificar las órdenes.

- _Gestión de los archivos del sistema:_ Se puede consultar el contenido de `usuarios.txt`, `menu.txt`, `pedidos.txt` y `bitacora.txt`, y anexarles notas con fecha directamente desde el programa.

- _Control de inactividad:_ Si pasan 10 minutos sin usar el sistema, se pregunta si se desea continuar en la sesión o regresar al inicio de sesión.

- _Resumen de ventas:_ El sistema puede leer `pedidos.txt` y mostrar cuántos pedidos se han hecho y el total vendido en el día.

- _Rutas de archivo consistentes:_ Los archivos `.txt` se localizan siempre a partir de la carpeta donde está el script, para que el programa funcione sin importar desde dónde se ejecute.

---

**Clasificación de Datos**

- `nombre_estudiante` = str = Nombre del estudiante al que pertenece la orden.

- `usuario` / `contrasena` = str = Datos de acceso del cajero, leídos desde `usuarios.txt`.

- `articulo` = str = Nombre del producto del menú.

- `categoria` = str = Categoría a la que pertenece un artículo (desayunos, bebida, hamburguesa, etc).

- `precio_articulo` = float = Precio del producto, usado para calcular el total.

- `cantidad` = int = Número de piezas de un artículo dentro de un pedido.

- `preparado` = boolean = Indica si un artículo está listo para servir sin espera o si requiere preparación.

- `menu` = list de dict = Colección de artículos disponibles, cada uno con su nombre, precio, disponibilidad y categoría.

- `pedido` = list de dict = Colección de artículos que un estudiante agregó a su orden.

- `total_pedido` = float = Resultado de sumar el subtotal (precio x cantidad) de todos los artículos del pedido.

- `fecha` = tuple = Fecha del día en formato (día, mes, año), usada en tickets, bitácora y notas.

- `ultima_actividad` = float = Marca de tiempo (timestamp) usada para medir la inactividad del sistema.

---

**Operadores de Lenguaje**

- Operadores matemáticos usados = `+`, `-`, `*` La suma se usa para el total del pedido y para acumular ventas del día, la resta se usa para retroceder un carácter al escribir la contraseña, la multiplicación se usa para calcular el subtotal de un artículo (precio x cantidad).

- Operadores Relacionales usados = `==`, `!=`, `<=`, `>=`, `<` "Igual a" se usa para validar usuario y contraseña e identificar artículos seleccionados, "Menor o igual a" y "Mayor o igual a" se usan para validar que una opción esté dentro del rango del menú y para controlar el límite de artículos y el tiempo de inactividad.

- Operadores lógicos usados = `and`, `or`, `not` Se usan en las validaciones de rango de opciones, en las condiciones para saber si un pedido requiere espera y para decidir si continuar o no en distintos menús del programa.

---

**Estructuras de Control**

- `if` `elif` `else` Se usan en el menú principal para dirigir cada opción, en las validaciones de usuario y contraseña, y para decidir si un artículo o categoría elegida es válida.

- `while` Se usa para mantener el programa activo hasta que el cajero decide salir, para repetir el login hasta que sea correcto, y para seguir agregando artículos a un pedido hasta que el estudiante termine o llegue al límite de 10.

- `for` Se usa para recorrer el menú y las categorías al mostrarlas en pantalla, para leer línea por línea los archivos `.txt`, y para controlar los intentos al validar si el usuario sigue activo tras la inactividad.

- `try` `except` Se usa en prácticamente todas las funciones que leen o escriben archivos, y en las que piden números al usuario, para evitar que el programa se cierre por un archivo faltante, datos inválidos o problemas de permisos.

**Diseño Algorítmico**

1. Al arrancar, se revisan y crean (si hacen falta) los archivos `usuarios.txt`, `menu.txt`, `pedidos.txt` y `bitacora.txt`.

2. Se muestra una pantalla de carga y se pide usuario y contraseña; no se deja pasar hasta que coincidan con `usuarios.txt`.

3. Ya dentro, se muestra un menú principal en bucle con las opciones: tomar pedido, agregar artículo, eliminar artículo, ver archivos del sistema, ver resumen de ventas y salir.

4. Al tomar un pedido, se pide el nombre del estudiante y se navega el menú por categoría y artículo, agregando piezas al pedido hasta que el estudiante decida terminar o se llegue al límite de 10 artículos.

5. Al cerrar el pedido, se calcula el total, se define si requiere tiempo de espera y se genera el ticket, que se muestra en pantalla y se guarda en `pedidos.txt`.

6. Cada vez que se modifica el menú (agregar o eliminar), el cambio se guarda en `menu.txt` y queda registrado en `bitacora.txt`.

7. Si el sistema detecta 10 minutos sin actividad, pregunta si se desea continuar; si no se responde correctamente después de 3 intentos, o si se responde que no, se cierra la sesión y se vuelve a pedir el login.

8. El programa sigue en el menú principal hasta que el usuario elige la opción de salir.

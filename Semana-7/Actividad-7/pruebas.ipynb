```python
def cargar_menu_desde_archivo():
    """Carga el menú desde menu.txt de forma segura."""
    menu_cargado = []
    try:
        with open("menu.txt", "r", encoding="utf-8") as f:
            for linea in f:
                partes = linea.strip().split(",")
                if len(partes) == 4:
                    articulo, precio, preparado_str, categoria = partes
                    menu_cargado.append({
                        "articulo": articulo,
                        "precio_articulo": float(precio),
                        "preparado": preparado_str.lower() == "true",
                        "categoria": categoria
                    })
    except FileNotFoundError:
        print("Aviso: menu.txt no existe, cargando menú vacío.")
    except Exception as e:
        print(f"Error al leer el menú: {e}")
    return menu_cargado

menu = cargar_menu_desde_archivo()
print(menu)
```

```python
import time
import os 
def pantalla_carga():
    """
    Función dedicada con pausa de carga de máximo 5 segundos 
    y aviso dinámico en consola.
    """
    print("\n[CARGANDO SISTEMA DE CAFETERÍA]")
    for segundo in range(1, 6):
        time.sleep(1)
        print(f"Cargando sistema... [{segundo}/5 segundos]")
    print("Carga del Sistema completado\n")


def iniciar_sesion():
    """
    Identificación de usuario y bienvenida dinámica solicitando nickname
    y aplicando operadores de cadenas.
    """
    pantalla_carga()
    sesion_iniciada = False
    
    while not sesion_iniciada:
        print("\n--- INICIO DE SESIÓN - CAFETERÍA TECMILENIO ---")
        nickname = input("Ingresa tu nombre o nickname: ").strip()
        
        # Bienvenida formal utilizando operadores de concatenación y formato de cadenas
        saludo = "¡Bienvenido/a al sistema, " + nickname.upper() + "! Acceso autorizado."
        decoracion = "*" * len(saludo)
        print(decoracion)
        print(saludo)
        print(decoracion)
        
        usuario_input = input("Usuario: ")
        contrasena_input = input("Contraseña: ")
        
        try:
            # Control de excepciones al leer el archivo de usuarios
            if not os.path.exists("usuarios.txt"):
                raise FileNotFoundError("El archivo usuarios.txt no se encuentra disponible.")
                
            encontrado = False
            with open("usuarios.txt", "r", encoding="utf-8") as archivo:
                for linea in archivo:
                    partes = linea.strip().split(",")
                    if len(partes) == 2:
                        user_db, pass_db = partes
                        if user_db == usuario_input and pass_db == contrasena_input:
                            encontrado = True
                            break
            
            if encontrado:
                sesion_iniciada = True
                print(f"Sesión iniciada correctamente por {nickname}.\n")
            else:
                print("Usuario o contraseña incorrectos. Intenta de nuevo.\n")
                
        except FileNotFoundError as fnf_error:
            print(f"{fnf_error}. Verifique los archivos del sistema.")
        except PermissionError:
            print("No hay permisos de lectura en el archivo de usuarios.")
        except Exception as ex:
            print(f"Error Inesperado: {ex}")

    return time.time()
iniciar_sesion()
```

```python
def inicializar_archivos_txt():
    """
    Verifica la existencia de los 4 archivos de texto obligatorios.
    Si no existen, los crea con datos iniciales por defecto.
    Controla excepciones de tipo FileNotFoundError y PermissionError amigablemente.
    """
    try:
        # 1. Archivo de usuarios
        if not os.path.exists("usuarios.txt"):
            with open("usuarios.txt", mode = "r", encoding="utf-8") as f:
                f.read()
                print(f)
        
        # 2. Archivo de menú
        if not os.path.exists("menu.txt"):
            with open("menu.txt", mode = "r", encoding="utf-8") as f:
                f.read()
                print(f)
        
        # 3. Archivo de pedidos
        if not os.path.exists("pedidos.txt"):
            with open("pedidos.txt", mode = "r", encoding="utf-8") as f:
                f.read()
                print(f)
        
        # 4. Archivo de bitácora
        if not os.path.exists("bitacora.txt"):
            with open("bitacora.txt", mode = "r", encoding="utf-8") as f:
                f.read()
                print(f)
                
    except FileNotFoundError as e:
        print(f"No se encontró la ruta del archivo inicial: {e}")
    except PermissionError as e:
        print(f"Error de permisos al intentar crear los archivos: {e}")
    except Exception as e:
        print(f"Ocurrió un error inesperado al inicializar archivos: {e}")


# Inicializamos los archivos al arrancar el programa
inicializar_archivos_txt()
```
